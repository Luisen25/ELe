```python
%%html
<!DOCTYPE html>
<html lang="es">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1, viewport-fit=cover">
<title>Modelo de tres poros: animación</title>
<style>
:root{--bg:#fff;--fg:#1d1d1f;--mut:#666;--card:#f3f4f6;--bl:rgba(255,120,120,.14);--pe:rgba(80,160,255,.14);--mem:#555;
box-sizing:border-box;padding-top:env(safe-area-inset-top,0px);padding-bottom:env(safe-area-inset-bottom,0px)}
@media (prefers-color-scheme:dark){:root:not([data-theme="light"]){--bg:#17181b;--fg:#ececec;--mut:#a0a0a0;--card:#24262b;--mem:#9a9a9a;--bl:rgba(255,100,100,.18);--pe:rgba(80,150,255,.18);color-scheme:dark}}
:root[data-theme="dark"]{--bg:#17181b;--fg:#ececec;--mut:#a0a0a0;--card:#24262b;--mem:#9a9a9a;--bl:rgba(255,100,100,.18);--pe:rgba(80,150,255,.18);color-scheme:dark}
html{scroll-padding-top:env(safe-area-inset-top,0px)}
body{margin:0;background:var(--bg);color:var(--fg);font-family:system-ui,-apple-system,"Segoe UI",sans-serif;padding:14px}
h1{font-size:1.15rem;margin:0 0 4px}
p{color:var(--mut);font-size:.85rem;margin:0 0 10px}
canvas{width:100%;max-width:900px;height:auto;display:block;border-radius:10px;background:var(--card)}
.row{display:flex;flex-wrap:wrap;gap:14px;margin:10px 0;max-width:900px;font-size:.85rem;align-items:center}
.row label{display:flex;flex-direction:column;gap:2px;min-width:200px}
.leg{display:flex;flex-wrap:wrap;gap:12px;max-width:900px;font-size:.82rem}
.leg span{display:flex;align-items:center;gap:5px}
.dot{width:11px;height:11px;border-radius:50%;display:inline-block}
button{padding:6px 12px;border-radius:8px;border:1px solid var(--mut);background:var(--card);color:var(--fg);cursor:pointer}
</style>
</head>
<body>
<h1>Modelo de tres poros: paso de agua y solutos</h1>
<p>Ilustración cualitativa: cada partícula se mueve al azar y solo cruza la membrana por un poro lo bastante grande.</p>
<canvas id="c" width="900" height="500"></canvas>
<div class="row">
  <label>Glucosa del dializado (empuje osmótico del agua)<input id="gl" type="range" min="0" max="100" value="40"></label>
  <label>Agitación térmica<input id="sp" type="range" min="30" max="200" value="100"></label>
  <button id="pl">Pausa</button><button id="rs">Reiniciar</button>
</div>
<div class="leg" id="leg"></div>
<script>
const cv=document.getElementById('c'),ctx=cv.getContext('2d'),W=900,H=500,MX=450,MW=8;
const pores=[{y:80,h:9,n:'Ultrapequeño (acuaporina)'},{y:250,h:24,n:'Pequeño'},{y:420,h:42,n:'Grande'}];
const T=[
 {k:'Agua',n:160,c:'#4aa3ff',r:3,ok:[0,1,2],side:0,st:1.0},
 {k:'Urea',n:45,c:'#2ca02c',r:5,ok:[1,2],side:-1,st:0.8},
 {k:'Glucosa',n:45,c:'#ff9f1c',r:6,ok:[1,2],side:1,st:0.7},
 {k:'Albúmina',n:10,c:'#d62728',r:11,ok:[2],side:-1,st:0.45}];
let P=[],run=true;
function rnd(a,b){return a+Math.random()*(b-a)}
function gauss(){return (Math.random()+Math.random()+Math.random()-1.5)*1.6}
function init(){P=[];T.forEach((t,ti)=>{for(let i=0;i<t.n;i++){
 const s=t.side===0?(Math.random()<.5?-1:1):t.side;
 P.push({t:ti,x:s<0?rnd(15,MX-30):rnd(MX+30,W-15),y:rnd(15,H-15),vx:0,vy:0})}})}
function passable(y,r,ok){return ok.some(i=>Math.abs(y-pores[i].y)<=pores[i].h-r*0.5)}
function step(){
 const sp=document.getElementById('sp').value/100,dr=document.getElementById('gl').value/100*0.9;
 for(const p of P){const t=T[p.t];
  p.vx=p.vx*.8+gauss()*t.st*sp*1.6;p.vy=p.vy*.8+gauss()*t.st*sp*1.6;
  let nx=p.x+p.vx+(t.k==='Agua'?dr:0),ny=p.y+p.vy;
  if(nx<t.r){nx=t.r;p.vx=Math.abs(p.vx)}if(nx>W-t.r){nx=W-t.r;p.vx=-Math.abs(p.vx)}
  if(ny<t.r){ny=t.r;p.vy=Math.abs(p.vy)}if(ny>H-t.r){ny=H-t.r;p.vy=-Math.abs(p.vy)}
  const inOld=Math.abs(p.x-MX)<MW+t.r*.3,inNew=Math.abs(nx-MX)<MW+t.r*.3;
  if(inNew&&!passable(ny,t.r,t.ok)){
   if(inOld){ny=p.y;p.vy=-p.vy}else{nx=p.x;p.vx=-p.vx}}
  p.x=nx;p.y=ny}}
function draw(){
 const cs=getComputedStyle(document.documentElement);
 ctx.clearRect(0,0,W,H);
 ctx.fillStyle=cs.getPropertyValue('--bl');ctx.fillRect(0,0,MX-MW,H);
 ctx.fillStyle=cs.getPropertyValue('--pe');ctx.fillRect(MX+MW,0,W-MX-MW,H);
 const fg=cs.getPropertyValue('--fg'),mem=cs.getPropertyValue('--mem');
 ctx.fillStyle=mem;let prev=0;
 [...pores].sort((a,b)=>a.y-b.y).forEach(o=>{ctx.fillRect(MX-MW,prev,MW*2,o.y-o.h-prev);prev=o.y+o.h});
 ctx.fillRect(MX-MW,prev,MW*2,H-prev);
 ctx.fillStyle=fg;ctx.font='12px system-ui';ctx.textAlign='center';
 ctx.fillText('CAPILAR (sangre)',MX/2,16);ctx.fillText('CAVIDAD PERITONEAL (dializado)',MX+MX/2,16);
 pores.forEach(o=>{ctx.fillText(o.n,MX,o.y-o.h-6)});
 for(const p of P){const t=T[p.t];ctx.fillStyle=t.c;ctx.beginPath();ctx.arc(p.x,p.y,t.r,0,6.283);ctx.fill()}
 const cnt=T.map((t,i)=>{const d=P.filter(p=>p.t===i&&p.x>MX).length;
  return `<span><i class="dot" style="background:${t.c}"></i>${t.k}: ${d}/${t.n} a la derecha</span>`}).join('');
 document.getElementById('leg').innerHTML=cnt}
function loop(){if(run)step();draw();requestAnimationFrame(loop)}
document.getElementById('pl').onclick=e=>{run=!run;e.target.textContent=run?'Pausa':'Continuar'};
document.getElementById('rs').onclick=init;
init();loop();
</script>
</body>
</html>
```
